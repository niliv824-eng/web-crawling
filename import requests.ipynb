```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
import time

driver_path = r"C:\Users\BiG\Downloads\chromedriver-win64\chromedriver-win64\chromedriver.exe"

service = Service(driver_path)
driver = webdriver.Chrome(service=service)

driver.maximize_window()

driver.get("https://www.digikala.com/search/?q=laptop")

time.sleep(5)

products = {}

for scroll in range(200):

    cards = driver.find_elements(By.TAG_NAME, "article")

    for card in cards:

        text = card.text.strip()

        if not text:
            continue

        lines = [x.strip() for x in text.split("\n") if x.strip()]

        if len(lines) < 2:
            continue

        # قیمت‌هایی که با عدد فارسی/انگلیسی هستند
        price = None

        for line in lines:
            cleaned = line.replace(",", "").replace("٬", "")

            if cleaned.isdigit():
                price = line

        # نام محصول
        name = lines[0]

        if price:
            # نام + قیمت را به عنوان کلید استفاده می‌کنیم
            key = (name, price)

            products[key] = {
                "name": name,
                "price": price
            }

    print(
        f"اسکرول {scroll + 1} | "
        f"محصولات یکتا: {len(products)}"
    )

    # اسکرول
    driver.execute_script(
        "window.scrollBy(0, 700);"
    )

    time.sleep(2)

print("\n==============================")
print("تعداد نهایی محصولات:", len(products))
print("==============================")

for i, product in enumerate(products.values(), 1):
    print(
        f"{i} | "
        f"{product['name']} | "
        f"{product['price']}"
    )

input("\nEnter برای خروج...")

driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
import time
import re

driver_path = r"C:\Users\BiG\Downloads\chromedriver-win64\chromedriver-win64\chromedriver.exe"

service = Service(driver_path)
driver = webdriver.Chrome(service=service)

driver.maximize_window()

driver.get("https://www.digikala.com/search/?q=laptop")

time.sleep(5)

products = {}

# تشخیص عدد فارسی
def is_price(text):
    text = text.replace(",", "").replace("٬", "").strip()

    # تبدیل اعداد فارسی به انگلیسی
    text = text.translate(str.maketrans(
        "۰۱۲۳۴۵۶۷۸۹",
        "0123456789"
    ))

    return text.isdigit() and len(text) >= 5


for scroll in range(300):

    cards = driver.find_elements(By.TAG_NAME, "article")

    for card in cards:

        lines = [
            x.strip()
            for x in card.text.split("\n")
            if x.strip()
        ]

        if not lines:
            continue

        # -------------------------
        # پیدا کردن قیمت‌ها
        # -------------------------

        prices = []

        for line in lines:

            if is_price(line):
                prices.append(line)

        if not prices:
            continue

        # -------------------------
        # اسم محصول
        # -------------------------

        name = lines[0]

        # -------------------------
        # قیمت نهایی
        # -------------------------

        # معمولاً آخرین قیمت موجود قیمت فعلی است
        price = prices[-1]

        # -------------------------
        # ذخیره محصول
        # -------------------------

        key = (name, price)

        products[key] = {
            "name": name,
            "price": price
        }

    print(
        f"اسکرول {scroll + 1} | "
        f"محصولات یکتا: {len(products)}"
    )

    # اسکرول
    driver.execute_script(
        "window.scrollBy(0, 700);"
    )

    time.sleep(2)


print("\n==============================")
print("تعداد محصولات:", len(products))
print("==============================")

for i, product in enumerate(products.values(), 1):

    print(
        f"{i} | "
        f"{product['name']} | "
        f"{product['price']}"
    )

input("\nبرای بستن Enter بزن...")

driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
import time

driver_path = r"C:\Users\BiG\Downloads\chromedriver-win64\chromedriver-win64\chromedriver.exe"

service = Service(driver_path)
driver = webdriver.Chrome(service=service)

driver.maximize_window()

driver.get("https://www.digikala.com/search/?q=laptop")

time.sleep(5)


def convert_farsi_number(text):
    table = str.maketrans(
        "۰۱۲۳۴۵۶۷۸۹",
        "0123456789"
    )
    return text.translate(table)


def is_price(text):
    text = text.strip()
    text = convert_farsi_number(text)
    text = text.replace(",", "").replace("٬", "")

    return text.isdigit() and len(text) >= 5


products = {}

last_count = 0
no_change = 0


for scroll in range(500):

    # کارت‌های فعلی DOM
    cards = driver.find_elements(By.TAG_NAME, "article")

    for card in cards:

        try:
            # اسم و قیمت را از خود کارت می‌گیریم
            lines = [
                x.strip()
                for x in card.text.split("\n")
                if x.strip()
            ]

            if not lines:
                continue

            # قیمت‌ها
            prices = []

            for line in lines:
                if is_price(line):
                    prices.append(line)

            if not prices:
                continue

            # اسم محصول
            name = lines[0]

            # قیمت فعلی
            price = prices[0]

            # حذف مواردی که محصول نیستند
            if len(name) < 10:
                continue

            # ذخیره
            key = (name, price)

            products[key] = {
                "name": name,
                "price": price
            }

        except Exception:
            continue


    # نمایش تعداد
    current_count = len(products)

    print(
        f"اسکرول {scroll + 1} | "
        f"قیمت‌های جمع‌شده: {current_count}"
    )


    # اگر محصول جدید پیدا نشد
    if current_count == last_count:
        no_change += 1
    else:
        no_change = 0

    last_count = current_count


    # اسکرول
    driver.execute_script(
        "window.scrollBy(0, 1000);"
    )

    time.sleep(1.5)


    # اگر مدت زیادی هیچ محصول جدیدی نیامد
    if no_change >= 20:
        print("محصول جدیدی پیدا نشد؛ پایان اسکرول.")
        break


print("\n==============================")
print("تعداد نهایی:", len(products))
print("==============================")


for i, product in enumerate(products.values(), 1):

    print(
        f"{i}. {product['name']}"
    )

    print(
        f"   قیمت: {product['price']}"
    )


input("\nEnter برای بستن...")

driver.quit()
```

```python
import asyncio
import re
from urllib.parse import urljoin

import pandas as pd
from playwright.async_api import async_playwright


# ---------------------------------------------------------
# Configuration
# ---------------------------------------------------------

URL = "https://www.amazon.de/s?k=laptop"
OUTPUT_FILE = "laptops.csv"

# تعداد صفحاتی که می‌خواهیم بخوانیم
MAX_PAGES = 3


# ---------------------------------------------------------
# Helper functions
# ---------------------------------------------------------

def clean_text(text):
    if not text:
        return ""
    return re.sub(r"\s+", " ", text).strip()


def extract_price(text):
    """
    تبدیل قیمت‌هایی مثل:
    549,99 €
    1.299,00 €
    €549.99
    به عدد float
    """
    if not text:
        return None

    text = text.replace("\xa0", " ").strip()

    # پیدا کردن قسمت عددی
    match = re.search(r"[\d.,]+", text)

    if not match:
        return None

    value = match.group(0)

    try:
        # فرمت آلمانی:
        # 1.299,99 -> 1299.99
        if "," in value:
            value = value.replace(".", "").replace(",", ".")
        else:
            # اگر فقط نقطه دارد، احتمالاً decimal انگلیسی است
            # ولی اگر چند نقطه وجود داشت، هزارگان در نظر می‌گیریم
            if value.count(".") > 1:
                value = value.replace(".", "")

        return float(value)

    except ValueError:
        return None


def extract_number(text):
    """
    استخراج عدد از متن review count
    مثال:
    '1.234' -> 1234
    '1234' -> 1234
    """
    if not text:
        return None

    match = re.search(r"[\d.,]+", text)

    if not match:
        return None

    value = match.group(0).replace(".", "").replace(",", "")

    try:
        return int(value)
    except ValueError:
        return None


# ---------------------------------------------------------
# Extract products from current page
# ---------------------------------------------------------

async def extract_products(page):
    products = []

    # Amazon search-result product cards
    cards = await page.locator(
        'div[data-component-type="s-search-result"]'
    ).all()

    print(f"Found {len(cards)} product cards")

    for card in cards:

        try:
            # ---------------------------------------------
            # ASIN
            # ---------------------------------------------

            asin = await card.get_attribute("data-asin")

            if not asin:
                continue

            # ---------------------------------------------
            # Title
            # ---------------------------------------------

            title = ""

            title_locator = card.locator(
                "h2 span"
            )

            if await title_locator.count() > 0:
                title = clean_text(
                    await title_locator.first.inner_text()
                )

            # fallback
            if not title:
                title_locator = card.locator("h2")

                if await title_locator.count() > 0:
                    title = clean_text(
                        await title_locator.first.inner_text()
                    )

            # ---------------------------------------------
            # Product URL
            # ---------------------------------------------

            url = ""

            link_locator = card.locator(
                'h2 a'
            )

            if await link_locator.count() > 0:
                href = await link_locator.first.get_attribute("href")

                if href:
                    url = urljoin(
                        "https://www.amazon.de",
                        href
                    )

            # ---------------------------------------------
            # Price
            # ---------------------------------------------

            price = None

            price_locator = card.locator(
                ".a-price .a-offscreen"
            )

            if await price_locator.count() > 0:

                price_text = await price_locator.first.inner_text()

                price = extract_price(price_text)

            # fallback
            if price is None:

                price_locator = card.locator(
                    ".a-price"
                )

                if await price_locator.count() > 0:

                    price_text = await price_locator.first.inner_text()

                    price = extract_price(price_text)

            # ---------------------------------------------
            # Rating
            # ---------------------------------------------

            rating = None

            rating_locator = card.locator(
                "span.a-icon-alt"
            )

            if await rating_locator.count() > 0:

                rating_text = await rating_locator.first.inner_text()

                match = re.search(
                    r"([0-5](?:[.,][0-9])?)",
                    rating_text
                )

                if match:
                    try:
                        rating = float(
                            match.group(1).replace(",", ".")
                        )
                    except ValueError:
                        pass

            # ---------------------------------------------
            # Review count
            # ---------------------------------------------

            review_count = None

            review_locator = card.locator(
                'a[href*="#customerReviews"]'
            )

            if await review_locator.count() > 0:

                review_text = await review_locator.first.inner_text()

                review_count = extract_number(review_text)

            # ---------------------------------------------
            # Image URL
            # ---------------------------------------------

            image_url = ""

            image_locator = card.locator(
                "img.s-image"
            )

            if await image_locator.count() > 0:

                image_url = await image_locator.first.get_attribute(
                    "src"
                )

                if not image_url:
                    image_url = await image_locator.first.get_attribute(
                        "data-src"
                    )

            # ---------------------------------------------
            # Availability
            # ---------------------------------------------

            availability = ""

            availability_locator = card.locator(
                ".a-color-price, .a-color-success"
            )

            if await availability_locator.count() > 0:

                availability = clean_text(
                    await availability_locator.first.inner_text()
                )

            # ---------------------------------------------
            # Save product
            # ---------------------------------------------

            product = {
                "asin": asin,
                "title": title,
                "price_eur": price,
                "rating": rating,
                "review_count": review_count,
                "availability": availability,
                "product_url": url,
                "image_url": image_url,
            }

            products.append(product)

        except Exception as e:
            print(f"Error extracting product: {e}")

    return products


# ---------------------------------------------------------
# Get next page
# ---------------------------------------------------------

async def get_next_page(page):

    next_button = page.locator(
        "a.s-pagination-next"
    )

    if await next_button.count() == 0:
        return False

    try:
        is_disabled = await next_button.get_attribute("aria-disabled")

        if is_disabled == "true":
            return False

        await next_button.click()

        await page.wait_for_load_state(
            "domcontentloaded"
        )

        # کمی صبر برای لود محصولات
        await page.wait_for_timeout(2000)

        return True

    except Exception as e:

        print(f"Could not go to next page: {e}")

        return False


# ---------------------------------------------------------
# Main scraper
# ---------------------------------------------------------

async def main():

    all_products = []

    async with async_playwright() as p:

        browser = await p.chromium.launch(
            headless=False
        )

        page = await browser.new_page(
            viewport={
                "width": 1440,
                "height": 900
            },
            locale="de-DE"
        )

        print("Opening Amazon...")

        try:

            await page.goto(
                URL,
                wait_until="domcontentloaded",
                timeout=60000
            )

        except Exception as e:

            print(f"Could not open page: {e}")

            await browser.close()

            return

        for page_number in range(1, MAX_PAGES + 1):

            print()
            print("=" * 50)
            print(f"PAGE {page_number}")
            print("=" * 50)

            # صبر برای render شدن محصولات
            await page.wait_for_timeout(3000)

            products = await extract_products(page)

            print(
                f"Extracted {len(products)} products"
            )

            all_products.extend(products)

            if page_number < MAX_PAGES:

                success = await get_next_page(page)

                if not success:
                    print(
                        "No more pages available."
                    )
                    break

        await browser.close()

    # -----------------------------------------------------
    # Remove duplicates
    # -----------------------------------------------------

    unique_products = {}

    for product in all_products:

        asin = product["asin"]

        if asin:
            unique_products[asin] = product

    products = list(unique_products.values())

    print()
    print(
        f"Total unique products: {len(products)}"
    )

    # -----------------------------------------------------
    # Create DataFrame
    # -----------------------------------------------------

    df = pd.DataFrame(products)

    if not df.empty:

        df = df[
            [
                "asin",
                "title",
                "price_eur",
                "rating",
                "review_count",
                "availability",
                "product_url",
                "image_url",
            ]
        ]

        df.to_csv(
            OUTPUT_FILE,
            index=False,
            encoding="utf-8-sig"
        )

        print()
        print(
            f"Dataset saved to: {OUTPUT_FILE}"
        )

        print()
        print(df.head())

    else:

        print(
            "No products were extracted."
        )


# ---------------------------------------------------------
# Run
# ---------------------------------------------------------

if __name__ == "__main__":
    asyncio.run(main())
```

```python
import re
from urllib.parse import urljoin

import pandas as pd
from playwright.async_api import async_playwright


# =========================================================
# تنظیمات
# =========================================================

URL = "https://www.amazon.de/s?k=laptop"
OUTPUT_FILE = "laptops.csv"

# تعداد صفحات
MAX_PAGES = 3


# =========================================================
# توابع کمکی
# =========================================================

def clean_text(text):
    if not text:
        return ""
    return re.sub(r"\s+", " ", text).strip()


def extract_price(text):
    if not text:
        return None

    text = text.replace("\xa0", " ").strip()

    match = re.search(r"[\d.,]+", text)

    if not match:
        return None

    value = match.group(0)

    try:
        # مثال:
        # 1.299,99 -> 1299.99
        if "," in value:
            value = value.replace(".", "")
            value = value.replace(",", ".")
        else:
            if value.count(".") > 1:
                value = value.replace(".", "")

        return float(value)

    except ValueError:
        return None


def extract_number(text):
    if not text:
        return None

    match = re.search(r"[\d.,]+", text)

    if not match:
        return None

    value = match.group(0)
    value = value.replace(".", "").replace(",", "")

    try:
        return int(value)
    except ValueError:
        return None


# =========================================================
# استخراج محصولات
# =========================================================

async def extract_products(page):

    products = []

    cards = await page.locator(
        'div[data-component-type="s-search-result"]'
    ).all()

    print(f"تعداد محصولات پیدا شده: {len(cards)}")

    for card in cards:

        try:

            # -------------------------------------------------
            # ASIN
            # -------------------------------------------------

            asin = await card.get_attribute("data-asin")

            if not asin:
                continue

            # -------------------------------------------------
            # Title
            # -------------------------------------------------

            title = ""

            locator = card.locator("h2 span")

            if await locator.count() > 0:
                title = clean_text(
                    await locator.first.inner_text()
                )

            if not title:
                locator = card.locator("h2")

                if await locator.count() > 0:
                    title = clean_text(
                        await locator.first.inner_text()
                    )

            # -------------------------------------------------
            # Product URL
            # -------------------------------------------------

            product_url = ""

            locator = card.locator("h2 a")

            if await locator.count() > 0:

                href = await locator.first.get_attribute("href")

                if href:
                    product_url = urljoin(
                        "https://www.amazon.de",
                        href
                    )

            # -------------------------------------------------
            # Price
            # -------------------------------------------------

            price = None

            locator = card.locator(
                ".a-price .a-offscreen"
            )

            if await locator.count() > 0:

                price_text = await locator.first.inner_text()

                price = extract_price(price_text)

            # -------------------------------------------------
            # Rating
            # -------------------------------------------------

            rating = None

            locator = card.locator(
                "span.a-icon-alt"
            )

            if await locator.count() > 0:

                rating_text = await locator.first.inner_text()

                match = re.search(
                    r"([0-5](?:[.,][0-9])?)",
                    rating_text
                )

                if match:

                    try:
                        rating = float(
                            match.group(1).replace(",", ".")
                        )
                    except ValueError:
                        pass

            # -------------------------------------------------
            # Reviews
            # -------------------------------------------------

            review_count = None

            locator = card.locator(
                'a[href*="#customerReviews"]'
            )

            if await locator.count() > 0:

                review_text = await locator.first.inner_text()

                review_count = extract_number(review_text)

            # -------------------------------------------------
            # Image
            # -------------------------------------------------

            image_url = ""

            locator = card.locator("img.s-image")

            if await locator.count() > 0:

                image_url = await locator.first.get_attribute(
                    "src"
                )

                if not image_url:

                    image_url = await locator.first.get_attribute(
                        "data-src"
                    )

            # -------------------------------------------------
            # Availability
            # -------------------------------------------------

            availability = ""

            locator = card.locator(
                ".a-color-price, .a-color-success"
            )

            if await locator.count() > 0:

                availability = clean_text(
                    await locator.first.inner_text()
                )

            # -------------------------------------------------
            # ذخیره
            # -------------------------------------------------

            products.append({

                "asin": asin,

                "title": title,

                "price_eur": price,

                "rating": rating,

                "review_count": review_count,

                "availability": availability,

                "product_url": product_url,

                "image_url": image_url

            })

        except Exception as e:

            print("خطا در استخراج محصول:", e)

    return products


# =========================================================
# رفتن به صفحه بعد
# =========================================================

async def go_to_next_page(page):

    next_button = page.locator(
        "a.s-pagination-next"
    )

    if await next_button.count() == 0:
        return False

    try:

        disabled = await next_button.get_attribute(
            "aria-disabled"
        )

        if disabled == "true":
            return False

        await next_button.click()

        await page.wait_for_load_state(
            "domcontentloaded"
        )

        await page.wait_for_timeout(3000)

        return True

    except Exception as e:

        print("خطا در رفتن به صفحه بعد:", e)

        return False


# =========================================================
# Scraper اصلی
# =========================================================

async def main():

    all_products = []

    async with async_playwright() as p:

        print("در حال باز کردن Chromium...")

        browser = await p.chromium.launch(
            headless=False
        )

        page = await browser.new_page(

            viewport={
                "width": 1440,
                "height": 900
            },

            locale="de-DE"

        )

        print("در حال باز کردن Amazon...")

        try:

            await page.goto(
                URL,
                wait_until="domcontentloaded",
                timeout=60000
            )

        except Exception as e:

            print("خطا در باز کردن Amazon:")
            print(e)

            await browser.close()

            return

        # =====================================================
        # صفحات
        # =====================================================

        for page_number in range(
            1,
            MAX_PAGES + 1
        ):

            print()
            print("=" * 60)
            print(f"PAGE {page_number}")
            print("=" * 60)

            await page.wait_for_timeout(3000)

            products = await extract_products(page)

            print(
                f"محصولات استخراج شده: {len(products)}"
            )

            all_products.extend(products)

            # صفحه بعد
            if page_number < MAX_PAGES:

                next_page = await go_to_next_page(page)

                if not next_page:

                    print(
                        "صفحه دیگری پیدا نشد."
                    )

                    break

        await browser.close()

    # =========================================================
    # حذف محصولات تکراری
    # =========================================================

    unique_products = {}

    for product in all_products:

        asin = product["asin"]

        if asin:
            unique_products[asin] = product

    products = list(
        unique_products.values()
    )

    print()
    print("=" * 60)
    print(
        f"تعداد محصولات نهایی: {len(products)}"
    )
    print("=" * 60)

    # =========================================================
    # DataFrame
    # =========================================================

    df = pd.DataFrame(products)

    if df.empty:

        print(
            "هیچ محصولی استخراج نشد."
        )

        return

    # ترتیب ستون‌ها

    columns = [

        "asin",
        "title",
        "price_eur",
        "rating",
        "review_count",
        "availability",
        "product_url",
        "image_url"

    ]

    df = df[columns]

    # =========================================================
    # ذخیره CSV
    # =========================================================

    df.to_csv(
        OUTPUT_FILE,
        index=False,
        encoding="utf-8-sig"
    )

    print()
    print(
        f"Dataset ذخیره شد: {OUTPUT_FILE}"
    )

    print()
    print("نمونه Dataset:")
    display(df.head(10))

    return df
```
