```python
import re
import pandas as pd

def extract_cpu(title):
    if pd.isna(title):
        return None

    patterns = [
        r'\bN\d{2,4}\b',                         # N95, N100, N150
        r'\bCeleron\b[^,|]*',                    # Celeron Quad core
        r'\bPentium\b[^,|]*',
        r'\bCore\s+i[3579]\b[^,|]*',             # Core i5, i7
        r'\bRyzen\s+[3579]\b[^,|]*',            # Ryzen 5, 7
        r'\bSnapdragon\s+X(?:\s+(?:Elite|Plus))?\b',
        r'\bApple\s+M[1-4](?:\s+(?:Pro|Max|Ultra))?\b',
    ]

    for pattern in patterns:
        match = re.search(pattern, title, re.IGNORECASE)
        if match:
            return match.group(0).strip()

    return None

    return None
mask = df["cpu"].isna()

df.loc[mask, "cpu"] = df.loc[mask, "title"].apply(extract_cpu)
df.loc[8, "cpu"] = "10100Y"
df.loc[76, "cpu"] = "3050E"
df.loc[213, "cpu"] = "Snapdragon X Elite"
df["cpu"].isna().sum()
```

```python
import pandas as pd

# خواندن دیتاست اصلی
df_original = pd.read_csv("laptops_amazon.csv")

# خواندن دیتاست آپدیت‌شده
df_updated = pd.read_csv("updated_data_laptops_amazon.csv")

# گرفتن فقط اطلاعات موردنیاز از دیتاست اصلی
source = df_original[["asin", "gaming", "touchscreen"]]

# حذف ستون‌های قبلی از دیتاست updated
df_updated = df_updated.drop(
    columns=["gaming", "touchscreen"],
    errors="ignore"
)

# اضافه کردن مقادیر جدید بر اساس ASIN
df_updated = df_updated.merge(
    source,
    on="asin",
    how="left"
)

# تبدیل True/False به 0/1
df_updated["gaming"] = df_updated["gaming"].astype(int)
df_updated["touchscreen"] = df_updated["touchscreen"].astype(int)

# ذخیره در یک فایل جدید
df_updated.to_csv(
    "updated_data_laptops_amazon_final.csv",
    index=False
)
```
