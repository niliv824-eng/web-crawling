```python
import pandas as pd
df=pd.read_csv("updated_data_laptops_amazon_final.csv")
categorical_cols = [
    "brand",
    "cpu",
    "gpu",
    "os",
    "keyboard",
    "resolution"
]

df_encoded = pd.get_dummies(
    df,
    columns=categorical_cols,
    dummy_na=True
)

print(df_encoded.head())
print(df_encoded.shape)
```
